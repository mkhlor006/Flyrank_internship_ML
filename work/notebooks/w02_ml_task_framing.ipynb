## 3. Success metric

*One metric you can defend. What number means 'good'?*



The primary success metric is **Precision@50**.

This measures how many of the top 50 pages identified by the model are actually declining pages. A higher Precision@50 means content teams spend their limited time reviewing pages that are more likely to benefit from investigation.

A useful model should improve Precision@50 compared with a simple rule-based approach.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
print("Positive pages:", (df["trend_direction"]=="down").sum())
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*



A simple rule such as "refresh every page with low CTR" would ignore other important factors such as impressions, search position, content age and combinations of these variables.

Machine learning can learn patterns across multiple observable features at the same time and identify combinations that may be associated with declining performance.

The model provides decision support based on measured patterns in the available data. It cannot prove why a page is declining or predict changes in Google's ranking algorithm.
